```python
import pandas as pd
from tabulate import tabulate
import re

def canonical_comment(s):
    s = str(s)

    # distance
    m = re.search(r"\bdistance\b.*?([0-9]+(?:\.[0-9]+)?)", s, re.I)
    if m:
        return f"distance = {float(m.group(1)):.1f}"

    # snr
    m = re.search(r"\bsnr\b.*?([0-9]+(?:\.[0-9]+)?)", s, re.I)
    if m:
        return f"snr = {float(m.group(1)):.1f}"

    return s

def to_fwf(df, fname):
    content = tabulate(df.values.tolist(), list(df.columns), tablefmt="plain")
    open(fname, "w").write(content)

pd.DataFrame.to_fwf = to_fwf
```

```python
c_band = pd.read_csv("../csv/maps_apr25_orig_c_band.csv")
c_band["comment"] = c_band["comment"].apply(canonical_comment)

s_band = pd.read_csv("../csv/maps_apr25_orig_s_band.csv")
s_band["comment"] = s_band["comment"].apply(canonical_comment)

x_band = pd.read_csv("../csv/maps_apr25_orig_x_band.csv")
x_band["comment"] = x_band["comment"].apply(canonical_comment)
```

```python
c_band.to_fwf("../csv/bad_images_c_band_ascii.txt")
s_band.to_fwf("../csv/bad_images_s_band_ascii.txt")
x_band.to_fwf("../csv/bad_images_x_band_ascii.txt")
```
