```python
from warnings import filterwarnings
from types import SimpleNamespace as sn
import pandas as pd
from yaml import load, FullLoader
from psycopg2 import connect
```

```python
filterwarnings("ignore")
with open("config.yml") as f:
    config = sn(**load(f, Loader=FullLoader))
config_db = sn(**config.db)

cnx = connect(
    host=config_db.host,
    dbname=config_db.dbname,
    user=config_db.user,
    password=config_db.psswd,
)
maps = pd.read_sql("select * from classes_t;", con=cnx)
```

```python
maps
```

```python
with open("val_paths.txt", "w") as f:
    for image, cls in zip(maps.file_name, maps.source_class):
        dir_name = image.split("_")[0]
        raw_file = f"{image.split('.')[0]}.fits"
        f.write(f"{dir_name}/{raw_file} {cls}\n")
```
